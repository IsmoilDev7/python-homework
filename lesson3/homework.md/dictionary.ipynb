1. Get Value: Given a dictionary and a key, retrieve the associated value, considering what to return if the key doesn’t exist.


```python
student = {
    'name': 'Adam',
    'age': 20,
    'city': 'New York'
}
key = 'name'
print(student.get(key))
```

```python
student = {
    'name': 'Adam',
    'age': 20,
    'city': 'New York'
}
key = 'name'
print("Is this key in the Dictionary:", key in student)
```
