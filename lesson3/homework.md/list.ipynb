1. Count Occurrences: Given a list and an element, find how many times the element appears in the list.

```python
numbers = [1, 2, 3, 2, 4, 2, 5, 1]
elements = int(input("enter element: "))
result = numbers.count(elements)
print("Counts", result)
```

11. Count Even Numbers: Given a list of integers, count how many of them are even.


```python
numbers9 = [3, 4, 5, 8, 5, 9, 12, 11, 16]
even_number = sum(map(lambda x: x % 2 == 0, numbers9))
print(even_number)
```

13. Find Sublist: Given a list and a sublist, check if the sublist exists within the list.


```python
numbers10 = [1, 2, 3, 4, 5, 6]
sublist = [3, 4, 5]
result8 = str(sublist)[1:-1] in str(numbers10)
print(result8)
```
