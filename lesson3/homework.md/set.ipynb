1. Union of Sets: Given two sets, create a new set that contains all unique elements from both sets.


```python
set1 = {1, 2, 3, 4}
set1_1 = {3, 4, 5, 6}
result1 = set1.union(set1_1)
print(result1)
```

2. Difference of Sets: Given two sets, create a new set with elements from the first set that are not in the second.


```python
set2 = {1, 2, 3, 4, 5, 6}
set2_2 = {4, 5, 6, 7, 8, 9}
result2 = set2.difference(set2_2)
print(result2)
```

3. Check Element: Given a set and an element, check if the element exists in the set.


```python
set3 = {3, 5, 2, 6, 7, 10}
element3 = 7
result3 = element3 in set3
print(result3)
```

4. Convert List to Set: Given a list, create a new set that contains only the unique elements from that list.


```python
list4 = [22, 11, 33, 44, 55, 66, 44, 11, 77, 66]
result4 = set(list4)
print(result4)
```

5. Remove Element: Given a set and an element, remove the element if it exists.


```python
set5 = {11, 23, 43, 53, 33, 25}
element5 = 43
set5.discard(element5)
print(set5)
```

6. Clear Set: Create a new empty set from an existing set.


```python
set6 = {}
print(set6)
```

7. Symmetric Difference: Given two sets, create a new set that contains elements that are in either set but not in both.


```python
set7 = {1, 2, 3, 4, 5}
set7_1 = {4, 5, 6, 7, 8}
result7 = set7.symmetric_difference(set7_1)
print(result7)
```
