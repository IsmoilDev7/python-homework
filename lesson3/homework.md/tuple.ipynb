1. Count Occurrences: Given a tuple and an element, find how many times the element appears in the tuple.


```python
tuple1 = (10, 20, 30, 20, 40, 20)
element1 = 20
print(tuple1.count(element1))
```

2. Max Element: From a given tuple, determine the largest element.


```python
import math
tuple1 = (10, 20, 30, 20, 40, 20)
print(max(tuple1))
```

3. Check Element: Given a tuple and an element, check if the element is present in the tuple.


```python
tuple3 = (10, 20, 30, 20, 40, 20)
num = 10
print("Is this num in tuple:", num in tuple3)
```

4. Last Element: Access the last element of a tuple, considering what to return if the tuple is empty.


```python
tuple4 = (10, 20, 30, 40, 50)
result = tuple4[-1:] or None
print(result)
```

5. Slice Tuple: Create a new tuple that contains only the first three elements of the original tuple.


```python
tuple5 = (10, 20, 30, 40, 50, 60)
result5 = tuple5[:3]
print(result5)
```

6. Check if Tuple is Empty: Determine if a tuple has any elements.


```python
tuple6 = (10, 20, 30)
print("Has element in the tuple:", bool(tuple6))
```

7. Find Second Largest: From a given tuple, find the second largest element.


```python
tuple7 = (2, 54, 23, 5, 25, 1, 3)
result7 = sorted(set(tuple7))[-2]
print(result7)
```

8. Create a Single Element Tuple: Create a tuple that contains a single specified element.


```python
tuple8 = (1,)
print(tuple8)
print(type(tuple8))
```

9. Check if Tuple is Sorted: Determine if the tuple is sorted in ascending order and return a boolean.


```python
tuple9 = (1, 2, 3, 5, 4)
result9 = tuple9 == tuple(sorted(tuple9))
print(result9)
```

10. Find Minimum of Subtuple: Given a tuple, find the minimum element of a specified subtuple.


```python
import math
tuple10 = (10, 25, 5, 40, 15, 30)
subtuple = tuple10[1:5]
result10 = min(subtuple)
print("Subtuple:", subtuple)
print("Minimal in the subtuple:", result10)
```
